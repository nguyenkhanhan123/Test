```python
# CELL 2: TẢI MẪU VRM + BLENDER TỰ ĐỔI MÀU ÁO HỎA ĐỎ + NẮM ĐẤM 10 NGÓN + RENDER DEMO HD
import os
%cd /content/Test
!git fetch origin main && git reset --hard origin/main

os.makedirs("/content/Test/assets/models_3d", exist_ok=True)
vrm_path = "/content/Test/assets/models_3d/hoa_do.vrm"
if not os.path.exists(vrm_path):
    !wget -q "https://raw.githubusercontent.com/pixiv/three-vrm/dev/packages/three-vrm/examples/models/VRM1_Constraint_Twist_Sample.vrm" -O {vrm_path}
print("VRM file:", vrm_path, f"({os.path.getsize(vrm_path)/1024/1024:.2f} MB)")

# Đổi tham số cuối để thử màu 5 chiến binh: hoa_do | thuy_lam | tho_vang | phong_luc | loi_hong
!/content/blender/blender --background --python vrm_demo_blender.py -- "/content/Test/assets/models_3d/hoa_do.vrm" "/content/Test/assets/models_3d/hoa_do_vrm_rigged.glb" "/content/Test/build_output/demo_vrm_hoa_do.mp4" hoa_do
```
